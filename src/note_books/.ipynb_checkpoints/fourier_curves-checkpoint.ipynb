# Fourier transform of every curve CSV in the project

Every per-text curve file (one CSV per text, named `factual_...` / `creative_...`) is read, and each text's curve is
put through `src/tools/fourier_transform.py`. Summary tables (e.g. `summary.csv`, the cluster tables) and the
`Qwen7B/archive` folders are left out.

For every folder, model and measured column you get:
1. **all texts' spectra on one graph**: thin line per text, thick line = factual / creative average
2. **a table**: per text, the strongest wave (period in words), fast share (share of word-to-word jitter), total power
3. **each text's own graph** (`SHOW_EACH_TEXT = True`): the curve with its smooth trend, and its spectrum

Each folder has its own cell below, so you can run only the ones you want.

```python
import sys
from collections import defaultdict
from pathlib import Path

import pandas as pd
from IPython.display import Markdown, display

REPO = next(folder for folder in [Path.cwd(), *Path.cwd().parents] if (folder / "src" / "tools").exists())
if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))

from src.tools.fourier_transform import as_curve, plot_fourier, plot_spectra, spectrum_table

SHOW_EACH_TEXT = True       # False: only the overlay graph and the table per folder / model / column
MIN_POINTS = 20             # curves shorter than this are skipped (too few points for a spectrum)

QWEN = REPO / "src" / "note_books" / "Qwen7B" / "results"
LCA_YOAV = REPO / "src" / "note_books" / "LCA-Yoav" / "results"

# every folder of curve CSVs: which columns are Fourier-transformed (any numeric column of the CSV can be added)
SOURCES = {
    "Results_CSV (LCA)":            dict(folder=REPO / "src" / "Results_CSV",           columns=["lca_height"]),
    "Results_CSV/cosine (BERT)":    dict(folder=REPO / "src" / "Results_CSV" / "cosine", columns=["mean_cos"]),
    "LCA-Yoav k5":                  dict(folder=LCA_YOAV / "k5",                         columns=["lca_height"]),
    "LCA-Yoav k5_clean":            dict(folder=LCA_YOAV / "k5_clean",                   columns=["lca_height"]),
    "Qwen7B run, words":            dict(folder=QWEN / "run" / "words",                  columns=["lca_height", "surprisal"],
                                         model="Qwen_Qwen2.5-7B-Instruct"),
    "Qwen7B run, tokens":           dict(folder=QWEN / "run" / "tokens",                 columns=["surprisal"],
                                         model="Qwen_Qwen2.5-7B-Instruct"),
    "Qwen7B burst run, words":      dict(folder=QWEN / "run_burst" / "words",            columns=["surprisal"],
                                         model="Qwen_Qwen2.5-7B-Instruct"),
    "Qwen7B manual run, words":     dict(folder=QWEN / "run_manual" / "words",           columns=["surprisal"],
                                         model="Qwen_Qwen2.5-7B-Instruct"),
}
NOT_CURVES = {"word_idx", "char", "block", "token_idx", "char_start", "char_end"}   # positions, not measurements


def curve_files(source):
    """[(model, text name, path)] of a source's per-text CSVs. 'model__text.csv' gives the model from the file
    name; otherwise the source's model. Files whose text name is not factual_ / creative_ are not curves."""
    files = []
    for path in sorted(Path(source["folder"]).glob("*.csv")):
        if "__" in path.stem:
            model, text = path.stem.split("__", 1)
        else:
            model, text = source.get("model", ""), path.stem
        if text.startswith(("factual", "creative")):
            files.append((model, text, path))
    return files


def numeric_columns(path):
    table = pd.read_csv(path, nrows=50)
    return [c for c in table.columns if c not in NOT_CURVES and pd.api.types.is_numeric_dtype(table[c])]


def fourier_source(name, columns=None, show_each=None):
    """Fourier-transform every text of one source: per model and column, the overlay graph, the table and
    (show_each) every text's own graph."""
    source = SOURCES[name]
    columns = columns or source["columns"]
    show_each = SHOW_EACH_TEXT if show_each is None else show_each
    by_model = defaultdict(list)
    for model, text, path in curve_files(source):
        by_model[model].append((text, path))

    for model, items in by_model.items():
        tables = {text: pd.read_csv(path) for text, path in items}
        for column in columns:
            curves = {}
            skipped = []
            for text, table in tables.items():
                if column in table.columns and len(as_curve(table, column)) >= MIN_POINTS:
                    curves[text] = table
                else:
                    skipped.append(text)
            display(Markdown(f"## {name} · {model or 'model not named'} · `{column}` ({len(curves)} texts)"))
            if skipped:
                print(f"skipped (missing column or fewer than {MIN_POINTS} points): {', '.join(skipped)}")
            if not curves:
                continue
            plot_spectra(curves, title=f"{name} · {model} · {column}: every text's spectrum", column=column)
            with pd.option_context("display.max_rows", 200, "display.width", 200):
                display(spectrum_table(curves, column=column))
            if show_each:
                for text, table in curves.items():
                    plot_fourier(table, title=f"{text}  ({model}, {column})", column=column)
```

```python
# What was found: every source, model, number of texts, and the columns that could be transformed
rows = []
for name, source in SOURCES.items():
    files = curve_files(source)
    for model in sorted({m for m, _t, _p in files}):
        paths = [p for m, _t, p in files if m == model]
        rows.append({"source": name, "model": model, "texts": len(paths),
                     "transformed columns": ", ".join(source["columns"]),
                     "other numeric columns": ", ".join(c for c in numeric_columns(paths[0])
                                                        if c not in source["columns"])})
inventory = pd.DataFrame(rows)
with pd.option_context("display.max_colwidth", 120):
    display(inventory)
print(f"{inventory['texts'].sum()} curve files")
```
