## סימנים לחלוקת חלון הקשר (Context Window) ב-NLP

###
* `.` סיום משפט מוחלט וסגירת רעיון שלם. -NLP, זהו העוגן המרכזי והחזק ביותר לאיפוס חלון ההקשר (Context Reset) ולחיתוך קלאסי של טקסט ליחידות משמעות עצמאיות.
  
* `,` מפריד פנימי רך (Soft Boundary). מסמן השהיה, פירוט רשימה או הפרדת פסוקיות (תנאי, תיאור), תוך שמירה על קשר הדוק לרעיון המרכזי העוטף. אינו שובר את חלון ההקשר אלא מארגן את התחביר בתוכו.
  
* `;` מפריד פנימי רך (Soft Boundary). מסמן השהיה, פירוט רשימה או הפרדת פסוקיות (תנאי, תיאור), תוך שמירה על קשר הדוק לרעיון המרכזי העוטף. אינו שובר את חלון ההקשר אלא מארגן את התחביר בתוכו.

* 
### סופי משפט מוחלטים (Sentence Boundaries)
סימנים שעוצרים לחלוטין את הרעיון הנוכחי ומתחילים אחד חדש:
* `?` **(סימן שאלה):** משנה את המבנה ומעיד על חוסר ודאות או בקשת מידע.
* `!` **(סימן קריאה):** מציין סיום משפט עם דגש רגשי או ציווי.

### מפרידי הרחבה או תפנית (Clause Separators)
סימנים שאחריהם הקונטקסט מקבל תפנית או הרחבה:
* `:` **(נקודתיים):** מסמנים פירוט, רשימה, ציטוט או מסקנה של מה שנאמר קודם.
* `-` או `—` **(קו מפריד):** מכניס הערת אגב או הסבר פתאומי באמצע משפט.
* `...` **(שלוש נקודות):** מסמנות מחשבה קטועה, השהיה או סיום פתוח (נפוץ בטקסט לא פורמלי).

### תת-הקשר (Enclosures & Sub-contexts)
סימנים שתוחמים קונטקסט שנמצא *בתוך* קונטקסט אחר:
* `()` או `[]` **(סוגריים):** תוחמים מידע משני. לעיתים קרובות מנותחים כחלון הקשר נפרד לחלוטין.
* `" "` או `' '` **(מירכאות):** ציטוט או דיבור ישיר לרוב מכילים קונטקסט שונה מהקונטקסט העוטף.

### סימני מבנה ועיצוב (Structural Markers)
תווים "שקופים" שקריטיים לעיבוד טקסט בקוד:
* `\n` או `\r\n` **(ירידת שורה):** מעבר פסקה מעיד כמעט תמיד על מעבר לנושא חדש.
* `\t` **(טאב):** מציין הפרדת נתונים מובנית.

---
**💡 טיפ לחיתוך (Chunking):**
הפילטר הבסיסי והנפוץ ביותר לחיתוך טקסט לחלונות הקשר הגיוניים ב-NLP מורכב מהסט הבא:
`[ . , ; : ? ! \n ]`


```python
def extract_context_segments(text):
    text_words = text.split()
    
    
    for i in range(1, len(text_words)):
        context = " ".join(text_words[:i])
        current_next_word = text_words[i]
        
    return 
```

```python
def create_top5_dict(valid_words, valid_embeddings, valid_probs):
    top_5_list = []
    for w, emb , org_prob in zip(valid_words, valid_embeddings , valid_probs):
        top_5_list.append([w, emb ,org_prob])  
    return top_5_list

def analyze_semantic_divergence(text):
    #TODO : change context window to read till spacial symbols
    text_words = text.split()
    embedded_dict = {}
     
    
    for i in range(1, len(text_words)):
        context = " ".join(text_words[:i])
        current_next_word = text_words[i]
        
        # clean the target word by removing punctuation and converting to lowercase
        clean_target_word = "".join([c for c in current_next_word if c.isalpha()]).lower()
        if not clean_target_word or clean_target_word in stop_words:
            continue

        batch = tokenizer(context, return_tensors="pt").to(model.device) #formatting to pytorch format
        with torch.no_grad():
            outputs = model(**batch, output_hidden_states=True)
            context_embedding = outputs.hidden_states[-1][0, -1, :]
            
            next_token_logits = outputs.logits[0, -1, :]
            top_probs, top_indices = torch.topk(F.softmax(next_token_logits, dim=-1), 50)

            valid_words = []
            valid_probs = []
            valid_embeddings = []


            for prob, idx in zip(top_probs, top_indices):
                word = tokenizer.decode(idx.item()).strip()
                # clean_predicted_word = "".join([c for c in word if c.isalpha()]).lower()
                has_special_letters = any(c.isalpha() for c in word)
                check_word = word.strip().lower()

                if has_special_letters and check_word not in stop_words:
                    valid_words.append(word)
                    valid_probs.append(prob.item())

                    token_embedding = model.get_input_embeddings()(idx).squeeze()
                    valid_embeddings.append(token_embedding)
                    # todo :
                    # check which cosine is stronger
                    # context embedeed vs top 5 prob embedded
                    # check each context window with the next context window (or current context window till next context window)

                    
            #todo: valid_word and make dict that compare cosine between 3 words to the ??
                
                if len(valid_words) == 5:
                    break

            # create embedded_dict
            target_token = tokenizer(current_next_word, return_tensors="pt").input_ids[0, -1].to(model.device)
            target_word_embedding = model.get_input_embeddings()(target_token).squeeze()

            cleaned_target_token = tokenizer(clean_target_word, return_tensors="pt").input_ids[0, -1].to(model.device)
            cleaned_target_word_embedding = model.get_input_embeddings()(target_token).squeeze()
            
            top_5_list = create_top5_dict(valid_words, valid_embeddings,valid_probs)
                
            batch_key = f"batch_{i}"
            embedded_dict[batch_key] = {
                "context_batch": context_embedding,
                "target_word": {current_next_word: target_word_embedding},
                "clean_target_word": {clean_target_word: cleaned_target_word_embedding},
                "top_5_prob" : top_5_list
            }
          
            print(f"(Context): '{context}'")
            # print(f"Context Embedding (first 5): {context_embedding[:5].tolist()}")
            print(f"(Target): '{current_next_word}' (clean_word: '{clean_target_word}')")
            print("topk 5 (filtered):")

            for word, prob, emb in zip(valid_words, valid_probs, valid_embeddings):
                print(f"  - {word}: {prob:.4f}")
            print("-" * 40)

    return embedded_dict


```

```python
with open('./texts_data/factual_wiki_computer_science.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
emb_dict = analyze_semantic_divergence(content)
```

```python
for batch_key, batch_data in emb_dict.items():
    print(f"\n=== {batch_key} ===")

    context_emb = batch_data["context_batch"]
    top_5_list = batch_data["top_5_prob"]
    target_word_dict = batch_data["target_word"]
    clean_target_dict = batch_data["clean_target_word"]
    
    target_word_str = list(target_word_dict.keys())[0]
    target_word_emb = list(target_word_dict.values())[0] 
    
    cleaned_target_word_str = list(clean_target_dict.keys())[0]
    cleaned_target_word_emb = list(clean_target_dict.values())[0] 
    
    print(f"Target word in text: '{target_word_str}'\n")
    print(f"{'Word':<15} | {'Context (Cos)':<15} | {'Target_token (Cos)':<15} | {'Target_clean (Cos)':<15} | {'Original Prob'}")
    print("-" * 65)
    
    for word, word_emb, prob in top_5_list:
        context_cosin = F.cosine_similarity(context_emb, word_emb, dim=0)
        target_cosin = F.cosine_similarity(target_word_emb, word_emb, dim=0)
        clean_target_cosin = F.cosine_similarity(target_word_emb, word_emb, dim=0)
        
        print(f"{word:<15} | {context_cosin.item():<15.4f} | {target_cosin.item():<15.4f} | {clean_target_cosin.item():<15.4f}| {prob:.4f}")
        
    
```
